```python
class Mart:
    def __init__(self, product, price, quantity, wallet):
        self.product = product
        self.price = price
        self.quantity = quantity
        self.wallet = wallet
        self.product_status = "not purchased"

    def cart(self):
        items = []
        items.append({self.product:self.price*self.quantity})
        self.product_status = "Not yet purchased"
        print("Product name:", items, "\n","Quantity: ", self.quantity, "\n","Status:", self.product_status)

    def shop(self):
        total_price = self.price*self.quantity
        if self.wallet >= total_price:
            self.wallet -= total_price
            self.product_status = "Purchased"
            print("Product: ", self.product)
            print("Price: ", self.price)
            print("Quantity: ", self.quantity)
            print("Status: ", self.product_status)
            print("Balance: ", self.wallet)
        else:
            print("Insufficient funds!")


c1 = Mart("iPhone 18 pro", 120000, 10, 10000000)

c1.shop()
```

```python
class Person:

    def __init__(self, personal_info, private_props, public_card_info):
        self.__personal_info = personal_info
        self._private_props = private_props
        self.public_card_info = public_card_info

    def public(self):
        print(self.public_card_info["name"])
        print(self.public_card_info["age"])
        print(self.public_card_info["work"])

    def private(self):
        print(self._private_props["land"])
        print(self._private_props["Profits by investments"])
        print(self._private_props["networth"])

    def personal(self):
        print(self.__personal_info["relationship"])
        print(self.__personal_info["friends"])
        print(self.__personal_info["close_friends"])
        print(self.__personal_info["family_members"])


    def details(self):
        inp = input("What you want to access\npress-1:Public_details\npress-2:Private_details\npress-3:Personal_details\n ")
        if inp == "1":
            self.public()

        elif inp == "2":
            self.private()

        elif inp == "3":
            pass1 = input("enter a password:")
            if pass1 == "password123":
                self.personal()
            else:
                print("Incorrect password")

        else:
            print("Invalid input")

class Status(Person):

    def __init__(self, personal_info, private_props, public_card_info, status):
        super().__init__(personal_info, private_props, public_card_info)
        self._status = status

    def status(self):
        print(f"{self.public_card_info['name']} have {self._status} Status.")



st1 = Status({"relationship":"3 GFs", "friends":"22 Friends", "close_friends":"4 besties", "family_members": "7 members in Family"}
    ,
    {"land": "120 acres-land", "Profits by investments": "$2000000 worth Investments", "networth": "$18000000000 of networth"},
    {"name":"Aaditya Sharma", "age":20, "work":"social activist"}, "Elite")

st1.details()
```

```python
class Calculator:

    def __init__(self, s1, s2):
        self.s1 = s1
        self.s2 = s2

    def cal(self):
        print(f"numbers: {self.s1}, {self.s2}")
        print(f"Addition: {self.s1. __radd__ (self.s2)}")
        print(f"Subtraction: {self.s1.__rsub__(self.s2)}")
        print(f"Multiplication: {self.s1.__rmul__ (self.s2)}")
        print(f"Division: {self.s1.__rtruediv__ (self.s2)}")
        print(f"Comparison(==): {self.s1.__eq__ (self.s2)}")
        print(f"Comparison(<=): {self.s1.__le__ (self.s2)}")
        print(f"Comparison(>): {self.s1.__gt__ (self.s2)}")
        print(f"Comparison(>=): {self.s1.__ge__ (self.s2)}")
        print(f"Comparison(!=): {self.s1.__ne__ (self.s2)}")

    def __iadd__(self, other):
        print(f"Addition_done: {str(self.s1)+ str(self.s2)}")

        return self

    def __str__(self):
        return f"the input numbers are: {self.s1}, {self.s2}"

    def __repr__(self):
        return f"num1: {self.s1}, num2: {self.s2}"

sum1 = Calculator(10,20)
sum1.cal()
a= Calculator(220, 2)
a += 10
a1 = a.s1
a.__str__()
a.__repr__()
```

```python
#coffee Machine proj:
print("Welcome to my SMART Coffee Machine!!")

menu1 = {"cappuccino":10, "latte": 8, "expresso": 11}
res = []

def resources(i1, i2, i3, m):
    coffee = i1
    milk = i2
    sugar = i3
    money = m

    print(F'''Report:
        Coffee: {coffee},
        Milk:{milk},
        Sugar: {sugar},
        Money: {money}''')

def menu(i1, i2, i3, m):

    user = input('''What would you like to have?
                1. Cappuccino --------$10
                2. Latte --------$8
                3. Expresso --------$11
                4. !!View Report!!''')
    admin = "pass123"
    res = [i1, i2, i3, m]
    if user == "1":
        if res[0] >= 2 and res[1] >= 3 and res[2] >= 6 and res[3] >= 10:
            print("Here is your tasty cappuccino!")
            res[0] = res[0] - 2
            res[1] = res[1] - 3
            res[2] = res[2] - 6
            choice(10, "cappuccino")
        else:
            print("sorry! Insufficient resources!")

    elif user == "2":
        if res[0] >= 4 and res[1] >= 2 and res[2] >= 4 and res[3] >= 8:
            print("Here is your tasty Latte!")
            res[0] = res[0] - 2
            res[1] = res[1] - 3
            res[2] = res[2] - 6
            choice(8, "latte")
        else:
            print("sorry! Insufficient resources!")

    elif user == "3":
        if res[0] >= 5 and res[1] >= 3 and res[2] >= 3 and res[3] >= 11:
            print("Here is your tasty Expresso!")
            res[0] = res[0] - 2
            res[1] = res[1] - 3
            res[2] = res[2] - 6
            choice(11, "expresso")
        else:
            print("sorry! Insufficient resources!")

    elif user == "4":
        adminpass = input("Enter password for admin access:")
        if adminpass == admin:
            resources(res[0], res[1], res[2], res[3])



def choice(total1, drink):
    print("Enter the amount used for transaction:")

    c5 = int(input("$5 coins: "))
    c10 = int(input("$10 coins: "))
    c20 = int(input("$20 coins: "))

    total = (c5 * 5) + (c10 * 10) + (c20 * 20)

    if total < total1:
        print("Sorry! Insufficient money!")
        print(f"Money refunded: {total}")
        goodbye()

    elif total > total1:
        change = total - total1
        print(f"Here is your change: {change}")
        print("here ia your bill:")
        bill(drink, menu1[drink])

    else:
        print("Transaction successful")
        print("here ia your bill:")
        bill(drink, menu1[drink])

    goodbye()

def bill(ing, pr, qt=1):

        print(f'''SMART Coffee Machine
            {ing}: {pr}*{qt}-----{pr*qt}
            Discount:---------{8/100*(pr*qt)}
            total:-------{(8/100*(pr*qt))-(pr*qt)}''')

        print(f"here is you cashback:{8/100*(pr*qt)}")

def goodbye():
    print("Thank you for visiting us!")
    print("visit again!")


cont = True
while cont:

    menu(100, 100, 100, 100)

    user1 = input("Do you want to continue? (y/n): ")
    if user1 == "y":
        cont = True
    else:
        cont = False




```
